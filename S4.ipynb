```python
#combining all concepts

students = [
    {"name": "Aarav", "study_hours": 6, "attendance": 85, "marks": 78},
    {"name": "Riya", "study_hours": 3, "attendance": 70, "marks": 55},
    {"name": "Kabir", "study_hours": 8, "attendance": 90, "marks": 35},
    {"name": "Meera", "study_hours": 2, "attendance": 60, "marks": 88}
]

for student in students:
    print(f"\nStudent: {student['name']}")

    # marks are more than 75 print performance strong
    if student['marks'] > 75:
        print("Performance: Strong in studies")

    # attendance is less than 75, print low attendance
    if student['attendance'] < 75:
        print("Attendance: Low attendance")

    # if student has studied for less than 4 hours, Warning needs more study time
    if student['study_hours'] < 4:
        print("Warning: Needs more study time")
```
