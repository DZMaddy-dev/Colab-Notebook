```python
sentance ="the movie is not good"
tokens=sentance.split()
hidden_state={
    "topic":None,
    "negative_active":False,
    "sentiment":None
}
for token in tokens:
  if token in ["movie","food","product"]:
    hidden_state["topic"]=token

  if token == "not":
    hidden_state["negative_active"]=True

  if token in ["good","bad","great","terrible"]:
    hidden_state["sentiment_word"]=token

  print("",token)
  print("",hidden_state)
```

```python
# creating a simple binary classifier nn
import tensorflow as tf
from tensorflow.keras import layers

# hyperparameters
vocab_size = 1000
embedding_dim = 16
sequence_length = 8

# starting of the simple rnn model
simple_rnn_model = tf.keras.Sequential([
    layers.Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=sequence_length),
    layers.SimpleRNN(32),
    layers.Dense(1, activation="sigmoid")
])

simple_rnn_model.summary()
```
