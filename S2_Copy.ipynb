```python
import pandas as pd
df=pd.read_csv("session4_dummy_students_data.csv")
print(df)
```

```python
print(df.shape)
```

```python
print(df.info())
```

```python
math_score=df['english'].to_numpy()
print(math_score)
print(type(math_score))
print(math_score.mean)
```

```python
df['average']=df[["math","science","english"]].mean(axis=1)
print(df[["name","math","science","english","average"]])
```

```python
top_student=df[df['average']>=85]
print(top_student[['name','average']])
```

```python
low_attendance=df[df['attendance_percent']<75]
print(low_attendance[['name','attendance_percent']])
```

```python
mathura=df[df['city']=='Mathura']
print(mathura['name'])
```

```python
import numpy as np
df['result'] = np.where(df['average'] >= 40, 'pass', 'fail')
df['low_assignment'] = np.where(df['assignments_submitted'] < 6, 'low assignment', 'perfect assignment')
print(df[['name', 'average', 'result', 'low_assignment']])
```
