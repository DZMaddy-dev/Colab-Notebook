```python
import pandas as pd
df=pd.read_csv('delhi_ncr_aqi_dataset.csv')
print(df.head())
```

```python
df.info()
```

```python
df=df.dropna()
print(df)
```

```python
df['date']=pd.to_datetime(df['date'])
df.info()
```

```python
df[['pm25','pm10','no2','so2','co','o3','aqi']].describe()
```

```python
df.groupby('date')['aqi'].mean().plot(figsize=(12,6))
```

```python
df.groupby('month')['aqi'].mean()
```

```python
df[['aqi','pm10']].corr()
```

```python
import matplotlib.pyplot as plt
df.plot(kind='scatter',x='pm25',y='aqi',alpha=0.5)
plt.show()
```
